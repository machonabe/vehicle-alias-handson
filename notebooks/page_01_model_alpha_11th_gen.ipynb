## 本文（以下をそのまま Page 本文に貼り付け）

### 概要

**Model Alpha 11th Gen** は、Alpha 系列の第11世代にあたる車両の共通業務定義です。日本・中国の販売、開発、生産では、それぞれ別の名称とコードで管理されています。このPageでは、それらが **同じ車両（共通機種ID `VEHICLE-001`）** を指すことを定義します。

- 共通機種ID：**VEHICLE-001**
- グローバル名称：**Model Alpha 11th Gen**
- 車両系列 / 世代：Alpha / 第11世代
- 含むパワートレイン：ガソリン（GAS）、ハイブリッド（HEV）
- 含まないもの：派生EV「Alpha e」（`VEHICLE-003`）、先代「Model Alpha 10th Gen」（`VEHICLE-002`）

- **確定値の正本は `vehicle_alias_master` テーブル**です。このPageの表は説明用の写しです。両者が食い違う場合は、`vehicle_alias_master` の `approval_status = 'APPROVED'` の行を正とします。食い違いはマスタ管理担当に報告してください。
- 候補（`approval_status = 'CANDIDATE'`）の行は、人が承認するまで集計に含めません。

### 同一機種と判断する条件

次をすべて満たす場合に、Model Alpha 11th Gen（VEHICLE-001）と判断します。詳細は Page「同一機種判定ルール」を参照してください。

1. コード（表記揺れを正規化したもの）が、上の対応表の確定コードと一致する。
2. そのコードの業務領域（販売・開発・生産）が一致する。
3. 対象データの年月が、そのコードの適用期間内にある。
4. 一致する共通機種IDが1つだけである。

### 同一機種と判断してはいけない条件

- **名称だけが一致する場合。**「Alpha」は先代（Model Alpha 10th Gen、コード JP-A10）でも使われています。
- **コードが似ているだけの場合。**例：`CN-X12` は `CN-X123`（Alpha）と `CN-X128`（Model Beta 3rd Gen）の両方に似ています。人が確認するまでは未確定です。
- **適用期間外の場合。**例：2025-06 以降の `CN-X123B`。
- **1つのコードが複数の共通機種IDに確定登録されている場合。**例：`MTO-990`。マスタの矛盾として扱い、集計から除外します。
- **AIの類似度スコアしか根拠がない場合。**スコアが高くても、人が承認するまでは候補です。

### 世代・販売期間・パワートレインが異なる場合の扱い

| ケース | 扱い | 例 |
|---|---|---|
| 世代が異なる | **別の共通機種ID** | Alpha 10th Gen（JP-A10、MTO-950）は VEHICLE-002 |
| 同一世代で、販売期間中のコード変更（システム移行など） | **同じ共通機種ID**。適用期間で旧コードと新コードを切り替える | CN-X123B（〜2025-05）→ CN-X123（2025-06〜） |
| 同一世代で、パワートレインやドアが異なる（GAS / HEV、5D） | **同じ共通機種ID**。パワートレインは販売データの属性列（`powertrain`）で区別する | JP-A11 の GAS と HEV |
| 同一世代でも、別の開発プロジェクト・開発コードを持つ派生車 | **別の共通機種ID** | Alpha e（JP-A11E、DEV-A11E、MTO-995）は VEHICLE-003 |
| 特別仕様車・追加グレード | 原則として同じ共通機種ID。ただし新コードは候補として登録し、人が承認してから確定する | JP-A11-SE |

### 集計・分析するときのルール

- 計画・販売・生産を横断して集計するときは **`v_vehicle_monthly_profitability`** を使います。粒度は「共通機種ID × 月」です。
- 元テーブル（`sales_actual`、`production_actual`、`development_plan`）をコードで直接結合しないでください。コード体系が異なるため、結合できないか、誤って結合されます。
- 集計から除外されたデータ（未解決・マスタ矛盾）は **`v_alias_review_queue`** で確認できます。回答するときは、除外されたデータがあることにも触れてください。
- 金額の単位は百万円、台数の単位は台です（ハンズオン用の架空データです）。
